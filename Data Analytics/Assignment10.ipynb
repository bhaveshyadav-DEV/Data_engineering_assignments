### Sales by Category

````python
%sql
SELECT
    category,
    SUM(total_revenue) AS total_sales
FROM dev.gold.sales_by_category
GROUP BY category
ORDER BY total_sales DESC;
````

### Best-Selling Products

````python
%sql
SELECT
    product_name,
    SUM(total_quantity_sold) AS total_quantity_sold
FROM dev.gold.best_selling_products
GROUP BY product_name
ORDER BY total_quantity_sold DESC
LIMIT 10;
````

### Top 10 Customers

````python
%sql
SELECT
    customer_id,
    SUM(total_sales) AS total_sales
FROM dev.gold.top_10_customers_view
GROUP BY
    customer_id
ORDER BY total_sales DESC
LIMIT 10;
````

# Task 8 — Serverless SQL Warehouse Performance Comparison

## Objective

The objective of this task is to understand the roles of Photon, Predictive I/O, and Intelligent Workload Management in improving the performance of a Databricks SQL Warehouse.

The comparison is used to justify why a Serverless SQL Warehouse is suitable for Cyntexa's AI/BI dashboard and Genie use case, where users need fast responses to interactive and ad-hoc queries.

---

# 1. Photon

## What is Photon?

Photon is a high-performance query execution engine designed to accelerate SQL and data processing workloads in Databricks.

It improves query execution performance by providing an optimized execution engine for SQL workloads.

## Role of Photon

Photon mainly focuses on:

* Faster query execution
* Improving SQL query performance
* Efficient processing of analytical workloads
* Improving performance for interactive SQL workloads

### Simple Example

When a user opens an AI/BI dashboard and runs a query such as:

```sql
SELECT
    region,
    SUM(revenue) AS total_revenue
FROM cyntexa_dev.gold.sales_summary
GROUP BY region;
```

Photon can help execute this analytical query efficiently.

---

# 2. Predictive I/O

## What is Predictive I/O?

Predictive I/O is a capability that helps improve data access efficiency by using information about the query and data to optimize I/O operations.

Its role is primarily related to efficiently accessing the data required by a query.

## Role of Predictive I/O

Predictive I/O focuses on:

* Improving data access
* Reducing unnecessary data reading
* Optimizing I/O operations
* Helping queries access required data efficiently

### Simple Example

If a dashboard query only needs a subset of a large dataset, efficient I/O can reduce unnecessary data access and help the query complete faster.

---

# 3. Intelligent Workload Management

## What is Intelligent Workload Management?

Intelligent Workload Management helps manage SQL workloads and available compute resources so that queries can receive appropriate resources.

It is focused more on workload and resource management rather than directly executing SQL operations.

## Role of Intelligent Workload Management

It helps with:

* Managing concurrent workloads
* Allocating resources to queries
* Handling changing workloads
* Maintaining good query performance under varying demand

### Simple Example

Suppose multiple users are using the dashboard at the same time while other users are asking Genie questions.

Intelligent Workload Management helps the warehouse manage these workloads and resources efficiently.

---

# 4. Comparison

| Technology                      | Primary Role                 | Main Focus                        | Benefit                                 |
| ------------------------------- | ---------------------------- | --------------------------------- | --------------------------------------- |
| Photon                          | Query execution engine       | SQL/query execution               | Faster query processing                 |
| Predictive I/O                  | Data access optimization     | I/O and data reading              | More efficient data access              |
| Intelligent Workload Management | Workload/resource management | Resource allocation and workloads | Better handling of concurrent workloads |

---

# 5. How They Work Together

These technologies address different parts of the query-processing problem.

```text
                    SQL Query
                       |
                       ↓
             Serverless SQL Warehouse
                       |
        ┌──────────────┼──────────────┐
        ↓              ↓              ↓
      Photon      Predictive I/O      IWM
        ↓              ↓              ↓
 Query execution   Data access    Workload/resource
  performance      efficiency       management
        │              │              │
        └──────────────┼──────────────┘
                       ↓
                Query Result
```

Photon focuses on query execution, Predictive I/O focuses on efficient data access, and Intelligent Workload Management focuses on managing workloads and resources.

Together, these capabilities support interactive SQL workloads.

# 7. Serverless SQL Warehouse

## Why use Serverless for this use case?

A Serverless SQL Warehouse is suitable for Cyntexa's dashboard and Genie use case because the workload is primarily interactive and analytical.

The main requirements are:

* Fast response for dashboard queries
* Support for ad-hoc SQL queries
* Support for Genie-generated queries
* Ability to handle changing workloads
* Reduced infrastructure management for the data team

Serverless is therefore a good fit for a workload where users may run queries at different times and query demand can change throughout the day.

---

# 8. Serverless vs Classic/Pro Warehouses

| Area                      | Serverless SQL Warehouse           | Classic/Pro Warehouse                  |
| ------------------------- | ---------------------------------- | -------------------------------------- |
| Infrastructure management | Lower management overhead          | More infrastructure management         |
| Interactive analytics     | Well suited                        | Can support it                         |
| Dashboard workloads       | Well suited                        | Can support it                         |
| Ad-hoc queries            | Well suited                        | Can support it                         |
| Genie workloads           | Well suited                        | Can support it                         |
| Scaling/management        | Designed for simplified management | Requires more configuration/management |
| Operational effort        | Lower                              | Higher                                 |

The choice depends on workload requirements, but for this assignment's interactive dashboard and Genie scenario, Serverless is the preferred option.

---

# 9. Why Serverless is Preferred for Cyntexa

Cyntexa's business stakeholders need self-service access to sales information through AI/BI dashboards and Genie.

The expected workload is:

```text
                    Business Users
                         |
              ┌──────────┴──────────┐
              ↓                     ↓
        AI/BI Dashboard          Genie
              ↓                     ↓
              └──────────┬──────────┘
                         ↓
               Serverless SQL Warehouse
                         ↓
                  Gold Sales Tables
```

This workload is interactive and may have changing query demand.

Serverless is preferred because it reduces infrastructure management while providing a suitable environment for interactive SQL analytics, dashboards, and Genie-driven questions.

---

# 10. Trade-off / Design Decision

The main reason for choosing Serverless is not simply that it is "faster."

The decision is based on the overall workload characteristics.

Cyntexa's use case requires interactive analytics, dashboard queries, and natural-language questions through Genie. These workloads can be unpredictable because different users may ask different questions at different times.

A Serverless SQL Warehouse is therefore a suitable choice because it provides a managed SQL analytics environment and reduces the operational effort required to manage infrastructure.

For this particular use case, the benefits of simplified management and support for interactive workloads outweigh the need to manage a warehouse infrastructure directly.

---

# 11. Final Recommendation

For the Cyntexa AI/BI dashboard and Genie use case, I recommend using a **Serverless SQL Warehouse**.

The reasoning is:

1. Photon supports efficient SQL query execution.
2. Predictive I/O helps optimize data access.
3. Intelligent Workload Management helps manage changing workloads and resources.
4. AI/BI dashboards require interactive query performance.
5. Genie generates SQL queries dynamically based on natural-language questions.
6. Serverless reduces infrastructure management overhead.
7. The workload is interactive and can vary depending on user demand.

Therefore, Serverless SQL Warehouse is an appropriate choice for Cyntexa's self-service analytics, AI/BI dashboard, and Genie use case.

---

# Conclusion

Photon, Predictive I/O, and Intelligent Workload Management address different aspects of SQL workload performance.

Photon focuses on query execution, Predictive I/O focuses on efficient data access, and Intelligent Workload Management focuses on workload and resource management.

For Cyntexa's AI/BI dashboard and Genie use case, a Serverless SQL Warehouse is a suitable choice because it supports interactive analytical workloads while reducing infrastructure management overhead.

The final decision should always be based on the actual workload, performance requirements, concurrency, and cost considerations.
