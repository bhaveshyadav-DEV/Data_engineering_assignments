# Task 3: Lakebase vs Delta Tables

## Objective

Read about Lakebase and write a short summary explaining when Lakebase should be used instead of a Delta table.

## Lakebase

Lakebase is suitable for **operational and transactional workloads** where applications need frequent, low-latency reads and writes.

Typical use cases include:

- Application transactions
- Order processing
- Inventory updates
- Application state and operational data
- Workloads that require frequent row-level changes

## Delta Tables

Delta tables are better suited for **analytical workloads** in the Databricks Lakehouse.

Typical use cases include:

- Large-scale data processing
- Aggregations
- Reporting
- Business intelligence (BI)
- Historical analysis
- Data engineering and analytical workloads

## When to Use Lakebase Instead of Delta

I would choose Lakebase when the primary requirement is an operational application that needs frequent and low-latency reads and writes.

For example, in an e-commerce application, Lakebase could be used to store current order status, inventory updates, and other operational information. Delta tables could then be used for downstream analytics such as sales trends, revenue analysis, customer analytics, and reporting.

Therefore, Lakebase and Delta tables can be used together rather than as direct replacements for each other:

**Lakebase → Operational / Transactional Workloads**

**Delta Tables → Analytical / OLAP Workloads**

## Production Architecture Example

```text
                 Application
                      |
                      v
              +---------------+
              |    Lakebase   |
              |     (OLTP)    |
              +---------------+
                      |
                      | Data for analytics
                      v
              +---------------+
              | Delta Tables  |
              |    Lakehouse  |
              +---------------+
                      |
                      v
              +---------------+
              | BI / Reports  |
              +---------------+
```

In this architecture, Lakebase handles the operational workload while Delta tables in the Databricks Lakehouse support downstream analytics and reporting.

## Final Summary

Lakebase is suitable for operational and transactional workloads where applications require frequent, low-latency reads and writes. I would choose Lakebase for use cases such as application transactions, order processing, inventory updates, or other OLTP workloads. Delta tables are better suited for analytical workloads involving large-scale data processing, aggregations, reporting, and BI. In a production architecture, Lakebase can handle the operational workload while Delta tables in the Databricks Lakehouse can be used for downstream analytics.

# Part D --- Document the two data-staleness scenarios

This is the most important written part of Task 4.

There are two concepts you need to explain:

1.  Scheduled refresh / Import mode
2.  Live query / DirectQuery mode

------------------------------------------------------------------------

## Scenario 1 --- Scheduled Refresh / Import Mode

In **Import mode**, Power BI stores a cached copy of the data in the
Power BI semantic model.

The data is not automatically updated every time the Databricks table
changes.

For example:

``` text
Databricks Gold Table
        ↓
   Data imported
        ↓
Power BI Semantic Model
        ↓
      Report
```

Suppose the Databricks table changes at:

``` text
10:00 AM
```

but Power BI's scheduled refresh is configured for:

``` text
2:00 PM
```

The Power BI report can continue showing the older data until the
refresh occurs.

### Why can data become stale?

Because the imported data remains cached until Power BI refreshes the
semantic model.

### How to reduce staleness

Configure a scheduled refresh in Power BI Service.

Typical path:

**Workspace → Semantic model → Refresh → Schedule refresh**

Configure the required frequency and time.

------------------------------------------------------------------------

## Scenario 2 --- Live Query / DirectQuery

In **DirectQuery mode**, Power BI does not import the source data into
the semantic model in the same way as Import mode.

Instead, queries are sent to the Databricks SQL Warehouse when Power BI
needs data.

Conceptually:

``` text
Power BI Report
      ↓
DirectQuery
      ↓
Databricks SQL Warehouse
      ↓
Databricks Table
```

Because queries go to the source, the report can reflect newer source
data without waiting for a normal full data-import refresh.

However, "live" does not mean that every visual is guaranteed to show an
instantaneous database state at every moment. Query execution, report
caching, semantic-model behavior, and other service behavior can affect
what the user sees.

### Why can data become stale?

DirectQuery generally does not depend on scheduled data import for the
source rows, but the displayed result can still be affected by
query/report caching or by the time at which the visual was last
queried.

Power BI documentation notes that DirectQuery/live-connection scenarios
generally don't rely on scheduled refresh for the source data.

6. Go back to your **Metastore settings** to verify that it now shows External OpenSharing: enabled.

# Step 2: Create Share, Add Table, and Setup Recipient (Host Account)
1. Open **Catalog Explorer** -> Click **Share** (top right) -> Select **OpenSharing**.
2. Go to the **Shared by me** tab and click the **Share data** button.
3. **Create Share:**
   * Enter **Share name** (e.g., gold_revenue_share).
   * Click **Save and continue**.
4. **Add Data Assets:**
   * Expand your catalog and schema (e.g., cyntexa_dev -> gold).
   * Select your table (e.g., daily_revenue_by_store).
   * Click **Save and continue**.
5. **Add Recipient:**
   * On the **Add recipients** step, click **Select a recipient** dropdown.
   * If recipient is not created, click **+ Create new recipient**:
     * Open your **Test Account** -> Go to **Catalog** -> **OpenSharing** -> Copy the **Sharing ID** from top right.
     * Paste the **Sharing ID** into the Host Account recipient creation box, enter recipient name (e.g., mohit_test), and save.
   * Select mohit_test from the dropdown list.
6. Click **Share data** at the bottom right.
7. You will now see your **Gold Table** successfully listed under the created Share (gold_revenue_share).

# Step 3: Configure Metastore Access Requests and Request OpenSharing Permissions (Test Account & Host Account)
1. **Verify Share Status (Host Account):**
   * Even though the table and recipient (mohit_test) are added in the Host Account's Share (gold_revenue_share), the shared data does not appear in the Test Account due to missing metastore access permissions.
2. **Configure Access Requests (Test Account):**
   * Open **Catalog Explorer** -> Click the **Gear icon** (top right) -> Select **Metastore** (metastore_aws_us_east_2).
   * In the **Details** tab, locate **Access requests** (currently showing Partially enabled).
   * Click the **Pencil (edit)** icon next to Partially enabled.
   * In the popup dialog under **Email**, enter the Host Account administrator's email address.
   * Click **Save**.
3. **Request OpenSharing Permissions (Test Account):**
   * Go back to **Catalog Explorer** -> **OpenSharing** -> **Shared with me** tab.
   * Click the **Request OpenSharing permissions** button.
   * In the request modal:
     * Select/Verify the **Principals** email address.
     * Ensure **USE PROVIDER** permission is checked.
     * Click **Request**.
4. **Approve Request (Host Account):**
   * The Host Account owner will receive an email titled **"A new Unity Catalog access request is awaiting your review"**.
   * Click **Open permission settings** inside the email and accept the request to enable sharing access.
