# Day 6 Assignment — Bronze Layer

## Task 1 — Create Bronze Table

The Bronze layer ingests the raw sales data as-is without applying business transformations.

All original columns and values are preserved, and an ingestion timestamp is added to track when the data was loaded into the Bronze layer.

Target table:

`dev.bronze.sales_raw1`

```python
sales_raw_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/dev/bronze/raw/sales.csv")
)
```

```python
from pyspark.sql.functions import current_timestamp
sales_bronze_df = sales_raw_df.withColumn(
    "ingestion_timestamp",
    current_timestamp()
)
```

```python
sales_bronze_df.printSchema()
```

```python
sales_bronze_df.write.mode("overwrite").saveAsTable("dev.bronze.sales_raw1")
```
